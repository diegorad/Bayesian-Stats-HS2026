```python
import scipy
import matplotlib.pyplot as plt
import numpy as np
```

```python
p = scipy.stats.norm(loc=0, scale=1)
q = scipy.stats.norm(loc=0, scale=1)
```

```python
grid = np.linspace(-2, 2, 100)
M=2
q_scaled=q.pdf(grid)*M
```

```python
plt.plot(grid, q_scaled)
plt.plot(grid, p.pdf(grid))
```

```python
sample_size = 1000
x_sample = np.random.uniform(size=sample_size)
```

```python
u = []
for sample in x_sample:
    u.append(np.random.uniform(low=0, high=q.pdf(sample)*M))
```

```python
accepted = []
for x, y in zip(x_sample, u):
    if(y < p.pdf(x)):
        accepted.append(x)
```

```python
len(accepted)
```

```python
plt.hist(np.array(accepted), density=True, alpha=0.5, label="Samples")
```
