```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
numpy_pi=np.pi
numpy_pi
```

```python
def pi_series(steps):
    sum = 0
    for k in range(steps):
        sum = sum + pow(-1, k)/(2*k+1)
    
    return 4*sum
```

```python
pi_series(10)
```

```python
series_results=[]

for steps in range(10, int(1e3), 1):
    val = pi_series(steps)
    series_results.append([steps, val])

series_results=np.array(series_results)
```

```python
plt.plot(series_results[:,0], [numpy_pi - val for val in series_results[:,1]], label="Series")
```

```python
def pi_mc(steps):
    approximation = []
    
    count = 0
    for step in range(int(steps)):
        sample = np.random.uniform(-1,1,2)
        if(sample[0]*sample[0] + sample[1]*sample[1] <= 1):
            count = count + 1
        approximation.append(4*count/steps)

    return 4*count/steps
```

```python
pi_mc(100)
```

```python
mc_results=[]

for steps in range(10, int(1e3), 1):
    val = pi_mc(steps)
    mc_results.append([steps, val])

mc_results=np.array(mc_results)
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

ax.plot(mc_results[:,0], [numpy_pi - val for val in mc_results[:,1]], label="Monte Carlo")
ax.plot(series_results[:,0], [numpy_pi - val for val in series_results[:,1]], label="Series")

ax.legend()

fig.show()
```
